```python
import sympy as sp
```

```python
# define symbols
m1, m2, k1, k2, k3, t = sp.symbols('m1 m2 k1 k2 k3 t')

x1 = sp.Function('x1')(t)
x2 = sp.Function('x2')(t)

# take derivatives
x1_dot = sp.diff(x1, t)
x1_ddot = sp.diff(x1_dot, t)
x2_dot = sp.diff(x2, t)
x2_ddot = sp.diff(x2_dot, t)

# Lagragian
T = 1/2*(m1*x1_dot**2 + m2*x2_dot**2)
V = 1/2*(k1*x1**2 + k2*(x2 - x1)**2 + k3*x2**2)
L = T - V
```

```python
# solve Euler-Lagrange for m1
eqn1 = sp.diff(sp.diff(L, x1_dot), t) - sp.diff(L, x1)
sln1 = sp.solve(eqn1, x1_ddot)[0]
sp.Eq(x1_ddot, sln1)
```

```python
# solve Euler-Lagrange for m2
eqn2 = sp.diff(sp.diff(L, x2_dot), t) - sp.diff(L, x2)
sln2 = sp.solve(eqn2, x2_ddot)[0]
sp.Eq(x2_ddot, sln2)
```
